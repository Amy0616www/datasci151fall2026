**Question (C)** Plotting functions with Numpy

<font size = "3">

Using Numpy arrays, create a **line plot** of the function $f(g(t))$, where

$$ f(x) = x^2 - 5x + 2 $$

$$ g(t) = \sin(t + 3) $$

at a set of 75 evenly spaced points in the interval $2\pi \leq t \leq 4\pi$.

```python
# SOLUTION 1
import numpy as np

t_vals = np.linspace(2*np.pi, 4*np.pi, 75)
x_vals = np.sin(t_vals + 3)

f_of_g_vals = x_vals**2 - 5 * x_vals + 2

plt.plot(t_vals, f_of_g_vals)
# Wouldn't require math notation formatting using "$" on a quiz
plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("Graph of $y = f(g(t))$ on interval $[2\\pi, 4\\pi]$")
plt.show()
```

```python
# SOLUTION 2
import numpy as np

start = 2*np.pi 
stop = 4*np.pi 
num_points = 75
t_vals = np.linspace(start, stop, num_points)
x_vals = np.sin(t_vals + 3)

f_of_g_vals = x_vals**2 - 5 * x_vals + 2

plt.plot(t_vals, f_of_g_vals)
# Wouldn't require math notation formatting using "$" on a quiz
plt.xlabel("$t$")
plt.ylabel("$y$")
plt.title("Graph of $y = f(g(t))$ on interval $[2\\pi, 4\\pi]$")
plt.show()
```

**Question (E)** Floating Point equality

<font size = "3">

The next code cell defines two variables (`x` and `y`) that in theory, should be equal. But when the cell is run, it prints out a message saying they are not the same. 

How would you choose a different comparison operation between `x` and `y` so that Python prints `True` instead?

```python
x = 0.1 * 7
y = 0.7

print("x and y are equal:", x == y)


# SOLUTION 1

# I choose delta = 10^-13. Something like 10^-14 or 10^-12 would be fine too
delta = 1e-13

print("x and y are equal:", abs(x - y) < delta)

```

```python
x = 0.1 * 7
y = 0.7

print("x and y are equal:", x == y)


# SOLUTION 2 - use intermediate Boolean variable

# I choose delta = 10^-13. Something like 10^-14 or 10^-12 would be fine too
delta = 1e-13

check_equal = abs(x - y) < delta

print("x and y are equal:", check_equal)

```
